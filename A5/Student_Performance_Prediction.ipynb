## Part D: Data Preprocessing

```
categorical_cols = ['Gender', 'InternetAccess', 'ExtraClasses', 'ParentEducation', 'PassStatus', 'RiskLevel']
numerical_cols = ['StudyHoursPerWeek', 'AttendancePercent', 'AssignmentScore', 'QuizAverage',
                   'PreviousGPA', 'SleepHours', 'ProjectScore']  # FinalScore held out as regression target

print("Categorical columns:", categorical_cols)
print("Numerical columns:", numerical_cols)

df_encoded = df.copy()
for col in categorical_cols:
    le = LabelEncoder()
    df_encoded[col + '_encoded'] = le.fit_transform(df_encoded[col])
    print(f"{col}: {dict(zip(le.classes_, le.transform(le.classes_)))}")

scaler_prep = StandardScaler()
scaled_array = scaler_prep.fit_transform(df_encoded[numerical_cols])
df_scaled_preview = pd.DataFrame(scaled_array, columns=[c + '_scaled' for c in numerical_cols])
print("\nScaled features (first 3 rows):")
display(df_scaled_preview.head(3))
```

**Why encoding and scaling matter here:** every model in this notebook (linear/logistic regression, KNN, PCA, LDA, t-SNE, gradient descent) relies on numeric math -- distances, dot products, or gradients -- so text categories like `"Good"`/`"Average"`/`"Poor"` must first become numbers (encoding). And since `AttendancePercent` (0-100) and `SleepHours` (roughly 4-10) live on very different scales, without `StandardScaler` the larger-range feature would dominate any distance- or gradient-based calculation regardless of its actual importance.

**Two target variables:** `FinalScore` (continuous, for regression in Part E) and `RiskLevel` (categorical, for classification in Part F). `PassStatus` was considered but rejected as a target due to its 698/2 imbalance.

```
df_model = df.copy()
for col in ['Gender', 'InternetAccess', 'ExtraClasses', 'ParentEducation']:
    df_model[col] = LabelEncoder().fit_transform(df_model[col])

# Regression feature set (target = FinalScore)
reg_feature_cols = ['StudyHoursPerWeek', 'AttendancePercent', 'AssignmentScore', 'QuizAverage',
                     'PreviousGPA', 'SleepHours', 'ProjectScore', 'Gender', 'InternetAccess',
                     'ExtraClasses', 'ParentEducation']
X_reg = df_model[reg_feature_cols]
y_reg = df_model['FinalScore']

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)
print("Regression split -- Train:", X_train_reg.shape, "Test:", X_test_reg.shape)
```

## Part E: Regression Task - Predict Final Score

```
scaler_reg = StandardScaler()
X_train_reg_scaled = scaler_reg.fit_transform(X_train_reg)
X_test_reg_scaled = scaler_reg.transform(X_test_reg)

reg_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42, max_depth=6),
    "Random Forest": RandomForestRegressor(random_state=42, n_estimators=200),
    "KNN": KNeighborsRegressor(n_neighbors=7),
}

reg_results = {}
reg_predictions = {}
for name, model in reg_models.items():
    model.fit(X_train_reg_scaled, y_train_reg)
    y_pred = model.predict(X_test_reg_scaled)
    reg_predictions[name] = y_pred
    mae = mean_absolute_error(y_test_reg, y_pred)
    mse = mean_squared_error(y_test_reg, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test_reg, y_pred)
    reg_results[name] = {"MAE": mae, "MSE": mse, "RMSE": rmse, "R2": r2}
    print(f"{name}: MAE={mae:.3f} MSE={mse:.3f} RMSE={rmse:.3f} R2={r2:.4f}")

reg_comparison = pd.DataFrame(reg_results).T.sort_values("R2", ascending=False)
print("\n===== Regression Model Comparison =====")
display(reg_comparison.round(4))
best_reg_name = reg_comparison.index[0]
print(f"Best regression model: {best_reg_name}")
```

```
fig, ax = plt.subplots(figsize=(7, 7))
best_pred = reg_predictions[best_reg_name]
ax.scatter(y_test_reg, best_pred, alpha=0.5, s=25, color='#4C72B0')
lims = [min(y_test_reg.min(), best_pred.min()), max(y_test_reg.max(), best_pred.max())]
ax.plot(lims, lims, 'r--', label='Perfect prediction')
ax.set_xlabel("Actual FinalScore")
ax.set_ylabel("Predicted FinalScore")
ax.set_title(f"Actual vs Predicted FinalScore ({best_reg_name})")
ax.legend()
plt.tight_layout()
plt.show()
```

```
# Which features help predict FinalScore
if best_reg_name == "Linear Regression":
    importances = sorted(zip(reg_feature_cols, np.abs(reg_models[best_reg_name].coef_)), key=lambda x: -x[1])
else:
    importances = sorted(zip(reg_feature_cols, reg_models[best_reg_name].feature_importances_), key=lambda x: -x[1])
print(f"Feature importance ({best_reg_name}):")
for feat, imp in importances:
    print(f"  {feat}: {imp:.3f}")
```

**Which features help predict FinalScore:** `StudyHoursPerWeek` dominates, followed by `PreviousGPA`, `QuizAverage`, and `AssignmentScore` -- consistent with the correlation analysis in Part A. Demographic/background features (`Gender`, `ParentEducation`, `InternetAccess`) contribute comparatively little.

**Explanation of model errors:** Linear Regression achieves the best R² (~0.69) and lowest error, suggesting `FinalScore` is close to a linear combination of the input features plus noise -- exactly the setting linear regression is built for. The tree-based/KNN models perform worse here (R² 0.44-0.62) likely because they're overfitting to noise in a relatively small (700-row) dataset with what appears to be a genuinely linear underlying relationship -- their flexibility to capture non-linear patterns isn't needed here and instead just adds variance. The remaining ~31% of unexplained variance (RMSE ≈ 4 points) likely reflects real-world randomness not captured by any of these features -- test anxiety, question difficulty variation, or other unmeasured factors.

## Part F: Classification Task - Predict Risk Level

```
# Target choice: RiskLevel (PassStatus rejected -- 698 vs 2, unusable)
print("RiskLevel distribution:", df['RiskLevel'].value_counts().to_dict())
print("PassStatus distribution:", df['PassStatus'].value_counts().to_dict(), "-- too imbalanced to use")

# Leakage check: does FinalScore trivially determine RiskLevel?
feature_cols_check = reg_feature_cols + ['FinalScore']
X_check = df_model[feature_cols_check]
y_risk = LabelEncoder().fit_transform(df_model['RiskLevel'])

dt_check = DecisionTreeClassifier(random_state=42, max_depth=3)
dt_check.fit(X_check, y_risk)
print("\nFeature importances (depth=3 tree, includes FinalScore):")
for feat, imp in sorted(zip(feature_cols_check, dt_check.feature_importances_), key=lambda x: -x[1])[:3]:
    print(f"  {feat}: {imp:.3f}")

print("\nFinalScore range by RiskLevel:")
print(df.groupby('RiskLevel')['FinalScore'].agg(['min', 'max']))
```

**Data leakage confirmed:** `FinalScore` alone gets a shallow tree to 100% importance/accuracy, and the min/max ranges show **zero overlap** between classes (Low Risk: 65.0-97.7, Medium Risk: 49.5-64.9) -- `RiskLevel` is a direct threshold on `FinalScore` (≥65 → Low Risk), not an independently meaningful label. `FinalScore` (and by extension anything derived purely from it) is excluded from the classification features below for an honest result.

```
clf_feature_cols = ['StudyHoursPerWeek', 'AttendancePercent', 'AssignmentScore', 'QuizAverage',
                     'PreviousGPA', 'SleepHours', 'ProjectScore', 'Gender', 'InternetAccess',
                     'ExtraClasses', 'ParentEducation']
X_clf = df_model[clf_feature_cols]

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_clf, y_risk, test_size=0.2, random_state=42, stratify=y_risk
)
scaler_clf = StandardScaler()
X_train_clf_scaled = scaler_clf.fit_transform(X_train_clf)
X_test_clf_scaled = scaler_clf.transform(X_test_clf)

print(f"Train: {X_train_clf.shape}, Test: {X_test_clf.shape}")
print(f"Class balance: {pd.Series(y_risk).value_counts().to_dict()} (0=Low Risk, 1=Medium Risk)")
```

```
clf_models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42, max_depth=6),
    "KNN": KNeighborsClassifier(n_neighbors=7),
    "Random Forest": RandomForestClassifier(random_state=42, n_estimators=200),
}

clf_results = {}
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
for ax, (name, model) in zip(axes, clf_models.items()):
    model.fit(X_train_clf_scaled, y_train_clf)
    y_pred = model.predict(X_test_clf_scaled)
    acc = accuracy_score(y_test_clf, y_pred)
    prec = precision_score(y_test_clf, y_pred)
    rec = recall_score(y_test_clf, y_pred)
    f1 = f1_score(y_test_clf, y_pred)
    cm = confusion_matrix(y_test_clf, y_pred)
    clf_results[name] = {"Accuracy": acc, "Precision": prec, "Recall": rec, "F1-score": f1}
    print(f"{name}: Acc={acc:.3f} Prec={prec:.3f} Rec={rec:.3f} F1={f1:.3f}")
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Low Risk", "Medium Risk"])
    disp.plot(ax=ax, cmap="Blues", colorbar=False)
    ax.set_title(name)
plt.tight_layout()
plt.show()

clf_comparison = pd.DataFrame(clf_results).T[["Accuracy", "Precision", "Recall", "F1-score"]].sort_values("F1-score", ascending=False)
print("\n===== Classification Model Comparison =====")
display(clf_comparison.round(4))
print(f"Best model (by F1): {clf_comparison.index[0]}")
```

**Which model is best for identifying students who need support:** Logistic Regression, with the best F1-score (~0.84) *and* the highest recall (~0.88) among the four. Recall is the critical metric for this use case: a false negative here means a student who genuinely needs support gets missed entirely, while a false positive just means a student gets offered support they may not strictly need -- a far lower-cost mistake. KNN's high precision (0.94) but poor recall (0.52) would be a bad choice operationally, since it would miss roughly **half** of the students actually at risk.

**Is accuracy alone enough for this problem?** No. `RiskLevel` is imbalanced (533 Low Risk vs 167 Medium Risk, roughly 76%/24%), so a model that always predicts "Low Risk" would already score ~76% accuracy while being completely useless for the department's actual goal of catching at-risk students. This is the same underlying issue that made `PassStatus` (698/2) entirely unusable as a target. Precision, recall, and F1 -- especially recall on the at-risk class -- are far more informative here than accuracy, since the practical cost of missing an at-risk student is much higher than the cost of flagging a student who turns out to be fine.

## Part G: Gradient Descent for Linear Regression (NumPy)

```
def gradient_descent_linreg(X, y, lr=0.01, n_iterations=1000):
    n_samples, n_features = X.shape
    X_b = np.column_stack([np.ones(n_samples), X])  # add bias column
    weights = np.zeros(n_features + 1)
    loss_history = []
    for i in range(n_iterations):
        y_pred = X_b @ weights
        error = y_pred - y
        loss = np.mean(error ** 2)  # MSE
        loss_history.append(loss)
        gradient = (2 / n_samples) * (X_b.T @ error)
        weights -= lr * gradient
    return weights, loss_history


weights, loss_history = gradient_descent_linreg(X_train_reg_scaled, y_train_reg.values, lr=0.05, n_iterations=500)

X_test_b = np.column_stack([np.ones(X_test_reg_scaled.shape[0]), X_test_reg_scaled])
y_pred_gd = X_test_b @ weights
mse_gd = mean_squared_error(y_test_reg, y_pred_gd)
r2_gd = r2_score(y_test_reg, y_pred_gd)
print(f"Gradient Descent Linear Regression: MSE={mse_gd:.3f} R2={r2_gd:.4f}")
print(f"Final training loss: {loss_history[-1]:.4f}")

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(loss_history)
ax.set_xlabel("Iteration")
ax.set_ylabel("Training Loss (MSE)")
ax.set_title("Gradient Descent Loss Curve (lr=0.05)")
plt.tight_layout()
plt.show()
```

```
# Compare with scikit-learn's closed-form Linear Regression (already trained in Part E)
sklearn_lr = reg_models["Linear Regression"]
print(f"scikit-learn Linear Regression: R2={reg_results['Linear Regression']['R2']:.4f}")
print(f"Gradient Descent Linear Regression: R2={r2_gd:.4f}")
print("\nWeight comparison (gradient descent vs sklearn), first 5 features:")
for i, feat in enumerate(reg_feature_cols[:5]):
    print(f"  {feat}: GD={weights[i+1]:.3f}, sklearn={sklearn_lr.coef_[i]:.3f}")
```

**Comparison with a traditional model:** the manually implemented gradient descent converges to essentially **identical** weights and R² as scikit-learn's closed-form `LinearRegression` -- strong validation that the from-scratch implementation is correct. This makes sense: for ordinary least-squares linear regression, gradient descent and the closed-form normal-equation solution are two different paths to the exact same optimum.

```
# Effect of learning rate on convergence
fig, ax = plt.subplots(figsize=(9, 5))
for lr_test in [0.001, 0.01, 0.05, 0.3]:
    _, lh = gradient_descent_linreg(X_train_reg_scaled, y_train_reg.values, lr=lr_test, n_iterations=200)
    ax.plot(lh, label=f"lr={lr_test}")
ax.set_xlabel("Iteration")
ax.set_ylabel("Training Loss (MSE)")
ax.set_title("Effect of Learning Rate on Convergence")
ax.set_ylim(0, 200)
ax.legend()
plt.tight_layout()
plt.show()

# Demonstrate divergence at high learning rates
print("Loss progression at high (unstable) learning rates:")
for lr_test in [1.0, 1.8]:
    _, lh = gradient_descent_linreg(X_train_reg_scaled, y_train_reg.values, lr=lr_test, n_iterations=6)
    print(f"  lr={lr_test}: {[round(l, 1) if np.isfinite(l) else l for l in lh]}")
```

**Why learning rate and iterations affect model performance:**
- **Learning rate too low** (e.g. 0.001): the loss curve barely moves in 200 iterations -- convergence is technically guaranteed but painfully slow, wasting compute and potentially never reaching a good solution within a practical iteration budget.
- **Learning rate well-tuned** (0.01-0.3 here): the loss drops sharply within the first ~20-30 iterations and settles near the optimum, matching the closed-form solution.
- **Learning rate too high** (1.0+): each update overshoots the minimum by more than the previous step, so the loss doesn't just fail to converge -- it **explodes** (visible above: loss goes from ~5,057 to hundreds of millions within 6 iterations). This is a direct consequence of the weight update step being proportional to the learning rate; too large a step size turns a descent into a diverging oscillation.
- **Iterations** control how much of the convergence curve you actually get to use: too few iterations stops training before the loss has flattened out (underfitting the optimization, not the data), while unnecessarily many iterations mostly just wastes compute once the loss has already plateaued.

## Part H: Dimensionality Reduction

```
# Reuse the honest (leakage-free) classification feature set and scaler from Part F
X_dr_scaled = scaler_clf.fit_transform(X_clf)
y_risk_labels = df['RiskLevel']

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_dr_scaled)
print("PCA explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance explained:", pca.explained_variance_ratio_.sum().round(3))

fig, ax = plt.subplots(figsize=(8, 6))
colors = {'Low Risk': '#4C72B0', 'Medium Risk': '#C44E52'}
for risk in ['Low Risk', 'Medium Risk']:
    mask = (y_risk_labels == risk).values
    ax.scatter(X_pca[mask, 0], X_pca[mask, 1], label=risk, alpha=0.5, s=25, color=colors[risk])
ax.set_xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)")
ax.set_ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)")
ax.set_title("PCA: Students Colored by Risk Level")
ax.legend()
plt.tight_layout()
plt.show()
```

```
# LDA -- supervised, so it directly uses RiskLevel to find the most separating direction.
# With only 2 classes present, LDA can produce at most 1 component (n_classes - 1 = 1).
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X_dr_scaled, y_risk)
print("LDA: 1 component (since only 2 RiskLevel classes are present in this dataset)")

fig, ax = plt.subplots(figsize=(9, 4))
for risk in ['Low Risk', 'Medium Risk']:
    mask = (y_risk_labels == risk).values
    ax.hist(X_lda[mask, 0], bins=30, alpha=0.6, label=risk, color=colors[risk])
ax.set_xlabel("LDA Component 1")
ax.set_ylabel("Count")
ax.set_title("LDA: Students Colored by Risk Level")
ax.legend()
plt.tight_layout()
plt.show()
```

```
tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_tsne = tsne.fit_transform(X_dr_scaled)

fig, ax = plt.subplots(figsize=(8, 6))
for risk in ['Low Risk', 'Medium Risk']:
    mask = (y_risk_labels == risk).values
    ax.scatter(X_tsne[mask, 0], X_tsne[mask, 1], label=risk, alpha=0.5, s=25, color=colors[risk])
ax.set_xlabel("t-SNE dimension 1")
ax.set_ylabel("t-SNE dimension 2")
ax.set_title("t-SNE: Students Colored by Risk Level")
ax.legend()
plt.tight_layout()
plt.show()
```

**Which method separates risk groups better, and why:** **LDA** separates the risk groups best, visibly and by construction. PCA and t-SNE are both *unsupervised* -- they organize the data purely by variance/neighborhood structure in the 11-feature space with no knowledge of `RiskLevel`, so the two colors end up substantially mixed in both plots (PCA explains only ~25% of total variance in 2D, offering little separation to begin with). LDA is *supervised*: it explicitly finds the axis that maximizes between-class separation relative to within-class spread, so even reduced to a single component, its histogram shows two clearly distinguishable (if slightly overlapping) peaks. The trade-off is that LDA's separation is somewhat "cheating" for visualization purposes -- it's allowed to look at the labels while PCA/t-SNE aren't -- so LDA is the right tool when the goal is confirming/understanding class separability, while PCA/t-SNE are more honest tools for exploring structure in unlabeled data.
